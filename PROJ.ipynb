```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import matplotlib.lines as mlines

fig, ax = plt.subplots(figsize=(12, 16))
ax.set_xlim(0, 10)
ax.set_ylim(0, 18)
ax.axis('off')

# Kolory
color_start = '#4CAF50'
color_process = '#2196F3'
color_model = '#FF9800'
color_output = '#9C27B0'
color_decision = '#F44336'

# Funkcja do rysowania prostokąta
def draw_box(ax, x, y, width, height, text, color, fontsize=11):
    box = FancyBboxPatch((x, y), width, height, 
                          boxstyle="round,pad=0.1", 
                          edgecolor='black', 
                          facecolor=color, 
                          linewidth=2)
    ax.add_patch(box)
    ax.text(x + width/2, y + height/2, text, 
            ha='center', va='center', 
            fontsize=fontsize, 
            fontweight='bold',
            color='white',
            wrap=True)

# Funkcja do rysowania strzałki
def draw_arrow(ax, x1, y1, x2, y2, label=''):
    arrow = FancyArrowPatch((x1, y1), (x2, y2),
                           arrowstyle='->,head_width=0.4,head_length=0.4',
                           color='black',
                           linewidth=2,
                           mutation_scale=20)
    ax.add_patch(arrow)
    if label:
        mid_x, mid_y = (x1 + x2) / 2, (y1 + y2) / 2
        ax.text(mid_x + 0.3, mid_y, label, 
                fontsize=9, 
                style='italic',
                bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray'))

# Funkcja do rysowania rombu (decyzja)
def draw_diamond(ax, x, y, width, height, text, color):
    diamond = mpatches.FancyBboxPatch((x, y), width, height,
                                      boxstyle="round,pad=0.05",
                                      transform=ax.transData,
                                      edgecolor='black',
                                      facecolor=color,
                                      linewidth=2)
    # Przybliżenie rombu przez obrócony kwadrat
    from matplotlib.patches import Polygon
    points = [
        [x + width/2, y + height],  # góra
        [x + width, y + height/2],  # prawo
        [x + width/2, y],           # dół
        [x, y + height/2]           # lewo
    ]
    diamond = Polygon(points, closed=True, 
                     edgecolor='black', 
                     facecolor=color, 
                     linewidth=2)
    ax.add_patch(diamond)
    ax.text(x + width/2, y + height/2, text,
            ha='center', va='center',
            fontsize=10,
            fontweight='bold',
            color='white')

# Tytuł
ax.text(5, 17.2, 'Schemat przebiegu działania aplikacji',
        ha='center', va='center',
        fontsize=16,
        fontweight='bold')

# 1. Start - Użytkownik wybiera obraz
draw_box(ax, 3, 15.7, 4, 0.8, 'START\nWybór obrazu', color_start, fontsize=11)
draw_arrow(ax, 5, 15.7, 5, 14.7)

# 2. Wczytywanie i przygotowanie obrazu
draw_box(ax, 3, 13.7, 4, 0.8, 'Wczytanie obrazu i przygotowanie danych', color_process, fontsize=11)
draw_arrow(ax, 5, 13.7, 5, 12.7)

# 3. Preprocessing
draw_box(ax, 3, 11.7, 4, 0.8, 'Preprocessing:\nzmiana rozmiaru, normalizacja', color_process, fontsize=11)
draw_arrow(ax, 5, 11.7, 5, 10.7)

# 4. Model YOLO - Detekcja
draw_box(ax, 3, 9.7, 4, 0.8, 'Wykonanie detekcji \n przez model YOLO', color_model, fontsize=11)
draw_arrow(ax, 5, 9.7, 5, 8.7)

# 5. Wyniki z modelu
draw_box(ax, 3, 7.7, 4, 0.8, 'Wyniki: ramki,\nklasy', color_model, fontsize=11)
draw_arrow(ax, 5, 7.7, 5, 6.7)

# 6. Połączenie bezpośrednie do outputów
draw_arrow(ax, 4.5, 6.7, 3.2, 6.1)
draw_arrow(ax, 5.5, 6.7, 7, 6.1)

# 7. Wyświetlenie informacji tekstowych
draw_box(ax, 0.5, 5.0, 4, 0.8, 'Wyświetlenie informacji:\nliczba uszkodzeń, klasy', color_output, fontsize=11)

# 8. Rysowanie ramek na obrazie
draw_box(ax, 5.5, 5.0, 4, 0.8, 'Rysowanie ramek\nna obrazie', color_output, fontsize=11)

# 9. Połączenie po wyświetleniu
draw_arrow(ax, 2.5, 5.0, 2.5, 4.1)
draw_arrow(ax, 7.5, 5.0, 7.5, 4.1)
draw_arrow(ax, 2.5, 4.1, 2.9, 4.1)
draw_arrow(ax, 7.5, 4.1, 7.1, 4.1)

# 10. Wyświetlenie wyniku końcowego
draw_box(ax, 3, 3.3, 4, 0.8, 'Wyświetlenie obrazu\nz zaznaczonymi uszkodzeniami', color_output, fontsize=11)
draw_arrow(ax, 5, 3.3, 5, 2.5)

# 11. Decyzja - Czy kontynuować?
draw_diamond(ax, 3.5, 1.8, 3, 0.6, 'Kolejna analiza', color_decision)

# Strzałka TAK - do czyszczenia (w dół)
draw_arrow(ax, 5, 1.8, 5, 1.2)
ax.text(5.2, 1.5, 'TAK', fontsize=9, style='italic',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray'))

# 12. Czyszczenie wyników
draw_box(ax, 3, 0.3, 4, 0.8, 'Czyszczenie wyników\ni reset interfejsu', color_process, fontsize=11)

# Strzałka powrotna do startu (po lewej stronie)
draw_arrow(ax, 3, 0.7, 0.3, 0.7)
draw_arrow(ax, 0.3, 0.7, 0.3, 16.1)
draw_arrow(ax, 0.3, 16.1, 2.9, 16.1)

# Strzałka NIE - koniec (w bok)
draw_arrow(ax, 6.5, 2.1, 8.4, 2.1)
ax.text(7.5, 2.3, 'NIE', fontsize=9, style='italic',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray'))
draw_box(ax, 8.5, 1.8, 1.2, 0.6, 'KONIEC', color_start, fontsize=11)

# Legenda
legend_elements = [
    mpatches.Patch(facecolor=color_start, edgecolor='black', label='Start/Koniec'),
    mpatches.Patch(facecolor=color_process, edgecolor='black', label='Przetwarzanie/Interakcja'),
    mpatches.Patch(facecolor=color_model, edgecolor='black', label='Model YOLO'),
    mpatches.Patch(facecolor=color_output, edgecolor='black', label='Wyświetlanie'),
    mpatches.Patch(facecolor=color_decision, edgecolor='black', label='Decyzja')
]
ax.legend(handles=legend_elements, loc='upper right', fontsize=9, framealpha=0.9)

plt.tight_layout()
plt.savefig('schemat_aplikacji_flowchart.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("Diagram przepływu został wygenerowany i zapisany jako 'schemat_aplikacji_flowchart.png'")
```

```python
# ...existing code...

from pycocotools.coco import COCO
import requests
import os
import json

# Ścieżka do pliku z adnotacjami COCO
annFile = 'annotations/instances_train2017.json'
coco = COCO(annFile)

# Pobierz ID kategorii dla samochodów
catIds = coco.getCatIds(catNms=['car'])
print(f"ID kategorii 'car': {catIds}")

# Pobierz ID zdjęć zawierających samochody
imgIds = coco.getImgIds(catIds=catIds)
print(f"Znaleziono {len(imgIds)} zdjęć z samochodami")

# Utwórz foldery
os.makedirs('car_dataset_yolo/images', exist_ok=True)
os.makedirs('car_dataset_yolo/labels', exist_ok=True)

def coco_to_yolo(bbox, img_width, img_height):
    """Konwertuje bbox z formatu COCO do YOLO"""
    x, y, w, h = bbox
    # YOLO format: center_x, center_y, width, height (wszystko znormalizowane)
    center_x = (x + w / 2) / img_width
    center_y = (y + h / 2) / img_height
    width = w / img_width
    height = h / img_height
    return center_x, center_y, width, height


# Pobierz zdjęcia i utwórz adnotacje YOLO
for i, imgId in enumerate(imgIds):  # Ograniczenie do 100 dla testu
    # Pobierz informacje o zdjęciu
    img_info = coco.loadImgs(imgId)[0]  # Dodaj [0] aby pobrać pierwszy element listy
    img_width = img_info['width']
    img_height = img_info['height']
    img_filename = img_info['file_name']
    
    # Pobierz zdjęcie
    try:
        img_url = img_info['coco_url']
        img_data = requests.get(img_url).content
        with open(f"car_dataset_yolo/images/{img_filename}", 'wb') as handler:
            handler.write(img_data)
    except Exception as e:
        print(f"Błąd przy pobieraniu {img_filename}: {e}")
        continue
    
    # Pobierz adnotacje dla tego zdjęcia (tylko samochody)
    annIds = coco.getAnnIds(imgIds=[imgId], catIds=catIds)
    anns = coco.loadAnns(annIds)
    
    # Utwórz plik z adnotacjami YOLO
    label_filename = img_filename.replace('.jpg', '.txt').replace('.png', '.txt')
    with open(f"car_dataset_yolo/labels/{label_filename}", 'w') as f:
        for ann in anns:
            bbox = ann['bbox']
            center_x, center_y, width, height = coco_to_yolo(bbox, img_width, img_height)
            # Klasa 0 dla samochodów (YOLO numeruje klasy od 0)
            f.write(f"0 {center_x:.6f} {center_y:.6f} {width:.6f} {height:.6f}\n")
    
    if (i + 1) % 10 == 0:
        print(f"Przetworzono {i + 1} zdjęć...")

# ...existing code...

# Utwórz plik classes.txt
with open('car_dataset_yolo/classes.txt', 'w') as f:
    f.write('car\n')

print(f"Dataset YOLO gotowy! Pobrano {min(len(imgIds), 100)} zdjęć z adnotacjami.")

# ...existing code...
```

```python

import yaml
import os
from pathlib import Path

# Przygotowanie struktury danych dla YOLOv5
def prepare_yolo_dataset():
    # Utwórz strukturę folderów
    os.makedirs('yolo_dataset/train/images', exist_ok=True)
    os.makedirs('yolo_dataset/train/labels', exist_ok=True)
    os.makedirs('yolo_dataset/val/images', exist_ok=True)
    os.makedirs('yolo_dataset/val/labels', exist_ok=True)
    
    # Przenieś pliki (80% train, 20% val)
    import shutil
    from glob import glob
    
    images = glob('car_dataset_yolo/images/*.jpg')
    split_idx = int(len(images) * 0.8)
    
    # Train set
    for img_path in images[:split_idx]:
        img_name = os.path.basename(img_path)
        label_name = img_name.replace('.jpg', '.txt')
        
        # Kopiuj obrazy i etykiety
        shutil.copy(img_path, f'yolo_dataset/train/images/{img_name}')
        if os.path.exists(f'car_dataset_yolo/labels/{label_name}'):
            shutil.copy(f'car_dataset_yolo/labels/{label_name}', f'yolo_dataset/train/labels/{label_name}')
    
    # Validation set
    for img_path in images[split_idx:]:
        img_name = os.path.basename(img_path)
        label_name = img_name.replace('.jpg', '.txt')
        
        # Kopiuj obrazy i etykiety
        shutil.copy(img_path, f'yolo_dataset/val/images/{img_name}')
        if os.path.exists(f'car_dataset_yolo/labels/{label_name}'):
            shutil.copy(f'car_dataset_yolo/labels/{label_name}', f'yolo_dataset/val/labels/{label_name}')

prepare_yolo_dataset()

# Utwórz plik konfiguracyjny dataset.yaml
dataset_config = {
    'path': os.path.abspath('yolo_dataset'),  # ścieżka do datasetu
    'train': 'train/images',  # train images
    'val': 'val/images',     # val images
    'nc': 1,                 # liczba klas
    'names': ['car']         # nazwy klas
}

with open('dataset.yaml', 'w') as f:
    yaml.dump(dataset_config, f)

print("Dataset przygotowany!")
```

```python
import torch
from ultralytics import YOLO
from codecarbon import EmissionsTracker
import os

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')
model = YOLO("yolo11n.pt")  

# Utwórz folder dla wyników CodeCarbon jeśli nie istnieje
os.makedirs("car_detection11n", exist_ok=True)

# Inicjalizacja CodeCarbon tracker
tracker = EmissionsTracker(
    project_name="yolo_car_detection_no_augment",
    output_dir="car_detection11n",
    output_file="emissions_no_augment.csv"
)

# Start śledzenia emisji
tracker.start()

try:
    model.train(
    data='dataset.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    patience=5,     # 0 = wyłączone early stopping, trenuj pełne 100 epok
    device=device,
    project='runs/train',
    name='car_detection11n'
)
finally:
    # Zatrzymanie trackera i wyświetlenie wyników
    emissions = tracker.stop()
    print(f"\n{'='*60}")
    print(f"CodeCarbon - Podsumowanie emisji CO2")
    print(f"{'='*60}")
    print(f"Całkowita emisja CO2: {emissions:.6f} kg")
    print(f"Wyniki zapisane w: car_detection2/emissions_no_augment.csv")
    print(f"{'='*60}\n")
```

```python
import matplotlib.pyplot as plt
import cv2
import os
import glob
import random

image_folder = 'yolo_dataset/train/images'    
label_folder = 'yolo_dataset/train/labels'   

n_samples = 10

label_files = glob.glob(os.path.join(label_folder, '*.txt'))
sampled_labels = random.sample(label_files, min(n_samples, len(label_files)))

class_names = {
    0: "samochod",
}

images_with_boxes = []

for label_path in sampled_labels:
    img_name = os.path.splitext(os.path.basename(label_path))[0]
    for ext in ['.jpg', '.jpeg', '.png', '.bmp']:
        img_path = os.path.join(image_folder, img_name + ext)
        if os.path.isfile(img_path):
            break
    else:
        print(f'Obrazek do {label_path} nie jest znaleziony!')
        continue
    
    image = cv2.imread(img_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) 

    h, w = image.shape[:2]

    with open(label_path, 'r') as f:
        for line in f:
            class_id, cx, cy, bw, bh = map(float, line.strip().split()[:5])
            class_id = int(class_id)
            x1 = int((cx - bw/2) * w)
            y1 = int((cy - bh/2) * h)
            x2 = int((cx + bw/2) * w)
            y2 = int((cy + bh/2) * h)
            color = (255, 0, 0)
            cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)
            cv2.putText(image, class_names.get(class_id, str(class_id)), (x1, y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2, cv2.LINE_AA)

    images_with_boxes.append((img_name, image))

fig, axes = plt.subplots(5, 2, figsize=(12, 20))
axes = axes.flatten()

for ax, (img_name, image) in zip(axes, images_with_boxes):
    ax.imshow(image)
    ax.axis('off')
    ax.set_title(img_name)

for ax in axes[len(images_with_boxes):]:
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# 📊 TABELA PORÓWNAWCZA METRYK WSZYSTKICH MODELI
import os
import pandas as pd
from IPython.display import display

print("📋 TWORZENIE TABELI PORÓWNAWCZEJ METRYK")
print("="*80)

# Ścieżki do wyników treningu
yolo8n_aug_path = "runs/train/car_detection3"  # YOLO8n z augmentacją
yolo8n_no_aug_path = "no_aug"  # YOLO8n bez augmentacji
yolo11n_path = "runs/train/car_detection11n"  # YOLO11n

# Funkcja do wczytania wyników z pliku results.csv
def load_training_results(path):
    results_file = os.path.join(path, "results.csv")
    if os.path.exists(results_file):
        df = pd.read_csv(results_file)
        df.columns = df.columns.str.strip()
        return df
    else:
        print(f"❌ Nie znaleziono pliku: {results_file}")
        return None

# Wczytaj wyniki dla wszystkich modeli
yolo8n_aug_results = load_training_results(yolo8n_aug_path)
yolo8n_no_aug_results = load_training_results(yolo8n_no_aug_path)
yolo11n_results = load_training_results(yolo11n_path)

if yolo8n_aug_results is not None and yolo8n_no_aug_results is not None and yolo11n_results is not None:
    # Pobierz ostatnie epoki
    yolo8n_aug_final = yolo8n_aug_results.iloc[-1]
    yolo8n_no_aug_final = yolo8n_no_aug_results.iloc[-1]
    yolo11n_final = yolo11n_results.iloc[-1]
    
    # Wczytaj emisje CO2
    yolo8n_aug_emissions = None
    yolo8n_no_aug_emissions = None
    yolo11n_emissions = None
    
    if os.path.exists("car_detection2/emissions_no_augment.csv"):
        yolo8n_aug_emissions_df = pd.read_csv("car_detection2/emissions_no_augment.csv")
        yolo8n_aug_emissions = yolo8n_aug_emissions_df['emissions'].iloc[-1]
    
    if os.path.exists("no_aug/emissions.csv"):
        yolo8n_no_aug_emissions_df = pd.read_csv("no_aug/emissions.csv")
        yolo8n_no_aug_emissions = yolo8n_no_aug_emissions_df['emissions'].iloc[-1]
    
    if os.path.exists("car_detection11n/emissions_no_augment.csv"):
        yolo11n_emissions_df = pd.read_csv("car_detection11n/emissions_no_augment.csv")
        yolo11n_emissions = yolo11n_emissions_df['emissions'].iloc[-1]
    
    # Funkcja pomocnicza do bezpiecznego pobrania wartości
    def safe_get_value(data, column, default=None):
        try:
            if column in data.index:
                return data[column]
            else:
                return default
        except:
            return default
    
    # Tworzenie słownika z metrykami
    metrics_dict = {
        'Metryka': [
            'Precision',
            'Recall',
            'mAP@50',
            'mAP@50-95',
            'Emisja CO2 (kg)'
        ],
        'YOLO8n (z aug.)': [
            safe_get_value(yolo8n_aug_final, 'metrics/precision(B)'),
            safe_get_value(yolo8n_aug_final, 'metrics/recall(B)'),
            safe_get_value(yolo8n_aug_final, 'metrics/mAP50(B)'),
            safe_get_value(yolo8n_aug_final, 'metrics/mAP50-95(B)'),
            yolo8n_aug_emissions
        ],
        'YOLO8n (bez aug.)': [
            safe_get_value(yolo8n_no_aug_final, 'metrics/precision(B)'),
            safe_get_value(yolo8n_no_aug_final, 'metrics/recall(B)'),
            safe_get_value(yolo8n_no_aug_final, 'metrics/mAP50(B)'),
            safe_get_value(yolo8n_no_aug_final, 'metrics/mAP50-95(B)'),
            yolo8n_no_aug_emissions
        ],
        'YOLO11n': [
            safe_get_value(yolo11n_final, 'metrics/precision(B)'),
            safe_get_value(yolo11n_final, 'metrics/recall(B)'),
            safe_get_value(yolo11n_final, 'metrics/mAP50(B)'),
            safe_get_value(yolo11n_final, 'metrics/mAP50-95(B)'),
            yolo11n_emissions
        ]
    }
    
    # Utwórz DataFrame
    comparison_df = pd.DataFrame(metrics_dict)
    
    # Formatowanie wartości liczbowych
    for col in ['YOLO8n (z aug.)', 'YOLO8n (bez aug.)', 'YOLO11n']:
        for idx in range(len(comparison_df)):
            if comparison_df.loc[idx, col] is not None:
                if idx == len(comparison_df) - 1:  # Emisja CO2
                    comparison_df.loc[idx, col] = f"{comparison_df.loc[idx, col]:.6f}"
                else:
                    comparison_df.loc[idx, col] = f"{comparison_df.loc[idx, col]:.4f}"
            else:
                comparison_df.loc[idx, col] = "N/A"
    
    # Wyświetl tabelę
    print("\n📊 TABELA PORÓWNAWCZA METRYK:")
    print("="*80)
    display(comparison_df)
    
    # Zapisz tabelę do CSV
    comparison_df.to_csv('model_comparison_metrics.csv', index=False, encoding='utf-8-sig')
    print("\n✅ Tabela zapisana do pliku: model_comparison_metrics.csv")
    
else:
    print("❌ Nie można wczytać wyników dla wszystkich modeli")
    print("Sprawdź czy foldery istnieją:")
    print(f"  - {yolo8n_aug_path}")
    print(f"  - {yolo8n_no_aug_path}")
    print(f"  - {yolo11n_path}")
```
